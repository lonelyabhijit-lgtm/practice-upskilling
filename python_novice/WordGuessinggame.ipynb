```python
import nltk
import ipywidgets as widgets
from IPython.display import display, clear_output
```

```python
nltk.download('wordnet')
from nltk.corpus import wordnet
import random
```

```python
word_len = int(input("Enter the length of the words"))
tot_words = int(input("Enter the number of words"))
words =[x for x in list(set([w.name().split('.')[0] for w in wordnet.all_synsets()])) if ('_' not in x)&(len(x)==word_len)]
random_words = random.sample(words, tot_words)
```

```python
for s in random_words:
    letters = [chr(i) for i in range(ord('a'), ord('z') + 1)]
    str_guess = s
    len_s = len(s)
    str_out = '_'*len_s
    int_try = 0
    while str_guess != '*'*len_s:
        print("Length of word", len_s)
        print("Letters left", letters)
        char_inp = input("Enter the character present in the word:")
        letters.remove(char_inp)
        if char_inp in s:
            str_guess = str_guess.replace(char_inp,'*')
            indices = [i+1 for i, x in enumerate(s) if x == char_inp]
            for i in indices:
                str_out = str_out[:i-1]+char_inp+str_out[i:]
            print("Word so far", str_out)
            print("Present in,",indices)
        else:
            print("Try again")
            print("Word so far", str_out)
        int_try += 1
    print("You have guessed the word %s in %d tries" %(s, int_try))
```
