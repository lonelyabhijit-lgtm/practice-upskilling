```python
try:    
    max_rows = int(input("Enter desired rows:"))
    if max_rows <= 0:
        print("Enter a valid positive integer")
    max_cols = max_rows * 2 - 1
    for i in range(max_rows):
        stars = " ".join("*" * (i+1))
        print(stars.center(max_cols))
except ValueError:
    print("Enter a valid positive integer")
```

```python
rows = int(input("Enter number of rows: "))
for i in range(1, rows + 1):
    if i == 1:
        stars = "*"
    elif i == rows:
        stars = " ".join("*" * i)
    else:
        stars = "*" + " " * (2 * i - 3) + "*"
    print(stars.center(rows * 2 - 1))
```

```python
import math

radius = int(input("Enter radius: "))
diameter = radius * 2
for y in range(diameter + 1):
    row = ""
    for x in range(diameter + 1):
        dist = math.sqrt((x - radius) ** 2 + (y - radius) ** 2)
        if radius - 0.5 < dist < radius + 0.5:
            row += "*"
        else:
            row += " "
    print(row)
```

```python
# Common emoji code points
emojis = [chr(128512), chr(128514), chr(10084), chr(128077), chr(127881), chr(128525), chr(128175), chr(128640), chr(127775), chr(128013)]

print("Emoji Code Points:")
print("-" * 50)
for emoji in emojis:
    code_point = ord(emoji)
    print(f"{emoji}  ->  {code_point:6d}  (0x{code_point:04X})")

print("\n" + "=" * 50)
print("Creating emojis from code points:")
print("=" * 50)

# Generate emojis from specific code points
code_points = [128512, 128151, 127881, 129412, 127775]
for cp in code_points:
    print(f"chr({cp}) = {chr(cp)}")
```

```python
# Default behavior - adds newline at end
print("Hello")
print("World")

print("\n" + "=" * 40 + "\n")

# Using end='' to print on same line
print("Hello", end='')
print("World")

print("\n" + "=" * 40 + "\n")

# Using end=' ' to separate with space
for i in range(5):
    print(i, end=' ')

print()  # Empty print to move to next line

print("\n" + "=" * 40 + "\n")

# Using custom end string
for char in "PYTHON":
    print(char, end='*')

print()  # Move to next line
```
